```python
# =====================================================================
# HETERODIFFROUTE-NET
# Complete Single-Cell Google Colab Implementation
# Five-Circuit Experimental Framework
# =====================================================================
#
# IMPORTANT:
# This implementation uses synthetic VLSI circuits.
# Results are proof-of-concept experimental results, not official
# ISPD/ICCAD benchmark results.
#
# =====================================================================

import os
import sys
import random
import heapq
import zipfile
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.nn.functional as F

from torch.utils.data import TensorDataset, DataLoader
from IPython.display import display

warnings.filterwarnings("ignore")

# =====================================================================
# 1. REPRODUCIBILITY AND DEVICE CONFIGURATION
# =====================================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("=" * 75)
print("HETERODIFFROUTE-NET")
print("VLSI PLACEMENT AND GLOBAL ROUTING")
print("=" * 75)

print("Execution Device:", DEVICE)

# =====================================================================
# 2. EXPERIMENTAL CONFIGURATION
# =====================================================================

NUM_NODES = 24
NUM_NETS = 36

GRID_SIZE = 32

TRAIN_SAMPLES = 120
TEST_SAMPLES = 5

GRAPH_EPOCHS = 25
CNN_EPOCHS = 20

BATCH_SIZE = 16
LEARNING_RATE = 0.001

ROUTING_CAPACITY = 2

OUTPUT_DIR = "/content/HeteroDiffRouteNet_Results"

os.makedirs(OUTPUT_DIR, exist_ok=True)

print("\nExperimental Configuration")
print("Training samples:", TRAIN_SAMPLES)
print("Testing circuits:", TEST_SAMPLES)
print("Number of nodes:", NUM_NODES)
print("Number of nets:", NUM_NETS)
print("Grid size:", GRID_SIZE)

# =====================================================================
# 3. SYNTHETIC VLSI CIRCUIT GENERATION
# =====================================================================

def generate_circuit():

    n = NUM_NODES

    # Reference coordinates

    true_positions = np.random.uniform(
        0.08, 0.92, (n, 2)
    ).astype(np.float32)

    # Component types:
    # 0 = Standard Cell
    # 1 = Macro
    # 2 = I/O Component

    node_types = np.random.randint(
        0, 3, n
    )

    widths = np.random.uniform(
        0.01, 0.04, n
    ).astype(np.float32)

    heights = np.random.uniform(
        0.01, 0.04, n
    ).astype(np.float32)

    # Noisy initial placement

    initial_positions = (
        true_positions
        + np.random.normal(
            0, 0.12, (n, 2)
        )
    )

    initial_positions = np.clip(
        initial_positions, 0.02, 0.98
    ).astype(np.float32)

    # Circuit connectivity

    edges = set()

    while len(edges) < NUM_NETS:

        a, b = np.random.choice(
            n, 2, replace=False
        )

        edges.add(
            tuple(sorted((int(a), int(b))))
        )

    edges = np.array(
        sorted(edges), dtype=np.int64
    )

    adjacency = np.eye(
        n, dtype=np.float32
    )

    for a, b in edges:

        adjacency[a, b] = 1
        adjacency[b, a] = 1

    degree = adjacency.sum(axis=1) - 1

    degree = degree / max(
        degree.max(), 1
    )

    type_features = np.eye(
        3, dtype=np.float32
    )[node_types]

    features = np.concatenate(
        [
            initial_positions,
            type_features,
            degree.reshape(-1, 1),
            widths.reshape(-1, 1),
            heights.reshape(-1, 1)
        ],
        axis=1
    ).astype(np.float32)

    return {
        "features": features,
        "adjacency": adjacency,
        "initial": initial_positions,
        "true": true_positions,
        "edges": edges,
        "types": node_types,
        "widths": widths,
        "heights": heights
    }


# =====================================================================
# 4. GENERATE TRAINING DATASET
# =====================================================================

print("\nGenerating training circuits...")

training_circuits = [
    generate_circuit()
    for _ in range(TRAIN_SAMPLES)
]

X_graph = np.array([
    c["features"]
    for c in training_circuits
])

A_graph = np.array([
    c["adjacency"]
    for c in training_circuits
])

Y_graph = np.array([
    c["true"]
    for c in training_circuits
])

print("Training dataset:", X_graph.shape)
print("Adjacency dataset:", A_graph.shape)
print("Placement targets:", Y_graph.shape)

# =====================================================================
# 5. HETEROGENEOUS GRAPH TRANSFORMER
# =====================================================================

class GraphTransformerBlock(nn.Module):

    def __init__(self, dimension=64, heads=4):

        super().__init__()

        self.heads = heads

        self.head_dim = dimension // heads

        self.qkv = nn.Linear(
            dimension, dimension * 3
        )

        self.projection = nn.Linear(
            dimension, dimension
        )

        self.norm1 = nn.LayerNorm(dimension)

        self.norm2 = nn.LayerNorm(dimension)

        self.ffn = nn.Sequential(
            nn.Linear(dimension, dimension * 2),
            nn.GELU(),
            nn.Linear(dimension * 2, dimension)
        )

    def forward(self, x, adjacency):

        batch, nodes, dimension = x.shape

        qkv = self.qkv(x)

        qkv = qkv.reshape(
            batch,
            nodes,
            3,
            self.heads,
            self.head_dim
        ).permute(2, 0, 3, 1, 4)

        q, k, v = qkv[0], qkv[1], qkv[2]

        scores = torch.matmul(
            q, k.transpose(-2, -1)
        ) / np.sqrt(self.head_dim)

        mask = adjacency.unsqueeze(1) <= 0

        scores = scores.masked_fill(
            mask, -1e4
        )

        attention = torch.softmax(
            scores, dim=-1
        )

        output = torch.matmul(
            attention, v
        )

        output = output.transpose(
            1, 2
        ).contiguous().reshape(
            batch, nodes, dimension
        )

        output = self.projection(output)

        x = self.norm1(x + output)

        x = self.norm2(
            x + self.ffn(x)
        )

        return x


class HeterogeneousGraphTransformer(nn.Module):

    def __init__(self):

        super().__init__()

        self.embedding = nn.Linear(
            8, 64
        )

        self.block1 = GraphTransformerBlock(
            64, 4
        )

        self.block2 = GraphTransformerBlock(
            64, 4
        )

        self.block3 = GraphTransformerBlock(
            64, 4
        )

        self.output = nn.Sequential(
            nn.Linear(64, 32),
            nn.GELU(),
            nn.Linear(32, 2),
            nn.Sigmoid()
        )

    def forward(self, features, adjacency):

        x = self.embedding(features)

        x = self.block1(x, adjacency)

        x = self.block2(x, adjacency)

        x = self.block3(x, adjacency)

        return self.output(x)


# =====================================================================
# 6. TRAIN GRAPH TRANSFORMER
# =====================================================================

graph_dataset = TensorDataset(
    torch.tensor(X_graph, dtype=torch.float32),
    torch.tensor(A_graph, dtype=torch.float32),
    torch.tensor(Y_graph, dtype=torch.float32)
)

graph_loader = DataLoader(
    graph_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True
)

graph_model = HeterogeneousGraphTransformer().to(DEVICE)

graph_optimizer = torch.optim.Adam(
    graph_model.parameters(),
    lr=LEARNING_RATE
)

graph_loss_fn = nn.MSELoss()

graph_losses = []

print("\nTraining Graph Transformer...")

for epoch in range(GRAPH_EPOCHS):

    graph_model.train()

    total_loss = 0

    for xb, ab, yb in graph_loader:

        xb = xb.to(DEVICE)
        ab = ab.to(DEVICE)
        yb = yb.to(DEVICE)

        graph_optimizer.zero_grad()

        prediction = graph_model(xb, ab)

        loss = graph_loss_fn(
            prediction, yb
        )

        loss.backward()

        graph_optimizer.step()

        total_loss += loss.item()

    avg_loss = total_loss / len(graph_loader)

    graph_losses.append(avg_loss)

    print(
        f"Graph Epoch {epoch+1:02d}/{GRAPH_EPOCHS} "
        f"| Loss: {avg_loss:.6f}"
    )


# =====================================================================
# 7. DIFFUSION-INSPIRED PLACEMENT REFINEMENT
# =====================================================================

def diffusion_refinement(
    predicted_positions,
    initial_positions,
    iterations=20
):

    coordinates = predicted_positions.copy()

    for step in range(iterations):

        alpha = 0.15 * (
            1 - step / iterations
        )

        # Iterative denoising-style refinement

        coordinates = (
            (1-alpha) * coordinates
            + alpha * predicted_positions
        )

        coordinates = np.clip(
            coordinates, 0.02, 0.98
        )

    return coordinates.astype(np.float32)


# =====================================================================
# 8. CONGESTION DEMAND MAP GENERATION
# =====================================================================

def generate_demand_map(positions, edges):

    demand = np.zeros(
        (GRID_SIZE, GRID_SIZE),
        dtype=np.float32
    )

    for a, b in edges:

        x1 = int(np.clip(
            positions[a, 0] * (GRID_SIZE-1),
            0, GRID_SIZE-1
        ))

        y1 = int(np.clip(
            positions[a, 1] * (GRID_SIZE-1),
            0, GRID_SIZE-1
        ))

        x2 = int(np.clip(
            positions[b, 0] * (GRID_SIZE-1),
            0, GRID_SIZE-1
        ))

        y2 = int(np.clip(
            positions[b, 1] * (GRID_SIZE-1),
            0, GRID_SIZE-1
        ))

        # Horizontal routing demand

        for x in range(
            min(x1, x2),
            max(x1, x2)+1
        ):

            demand[y1, x] += 1

        # Vertical routing demand

        for y in range(
            min(y1, y2),
            max(y1, y2)+1
        ):

            demand[y, x2] += 1

    return demand


# =====================================================================
# 9. GENERATE CNN TRAINING DATA
# =====================================================================

print("\nGenerating congestion dataset...")

cnn_inputs = []
cnn_targets = []

for circuit in training_circuits:

    initial_map = generate_demand_map(
        circuit["initial"],
        circuit["edges"]
    )

    target_map = generate_demand_map(
        circuit["true"],
        circuit["edges"]
    )

    cnn_inputs.append(initial_map)

    cnn_targets.append(target_map)

cnn_X = torch.tensor(
    np.array(cnn_inputs)[:, None],
    dtype=torch.float32
)

cnn_Y = torch.tensor(
    np.array(cnn_targets)[:, None],
    dtype=torch.float32
)

cnn_dataset = TensorDataset(
    cnn_X, cnn_Y
)

cnn_loader = DataLoader(
    cnn_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True
)


# =====================================================================
# 10. CNN CONGESTION PREDICTION MODEL
# =====================================================================

class CongestionCNN(nn.Module):

    def __init__(self):

        super().__init__()

        self.network = nn.Sequential(

            nn.Conv2d(
                1, 16, 3, padding=1
            ),

            nn.ReLU(),

            nn.Conv2d(
                16, 32, 3, padding=1
            ),

            nn.ReLU(),

            nn.Conv2d(
                32, 16, 3, padding=1
            ),

            nn.ReLU(),

            nn.Conv2d(
                16, 1, 1
            ),

            nn.Softplus()
        )

    def forward(self, x):

        return self.network(x)


# =====================================================================
# 11. TRAIN CNN
# =====================================================================

cnn_model = CongestionCNN().to(DEVICE)

cnn_optimizer = torch.optim.Adam(
    cnn_model.parameters(),
    lr=LEARNING_RATE
)

cnn_loss_fn = nn.MSELoss()

cnn_losses = []

print("\nTraining CNN Congestion Predictor...")

for epoch in range(CNN_EPOCHS):

    cnn_model.train()

    total_loss = 0

    for xb, yb in cnn_loader:

        xb = xb.to(DEVICE)
        yb = yb.to(DEVICE)

        cnn_optimizer.zero_grad()

        prediction = cnn_model(xb)

        loss = cnn_loss_fn(
            prediction, yb
        )

        loss.backward()

        cnn_optimizer.step()

        total_loss += loss.item()

    avg_loss = total_loss / len(cnn_loader)

    cnn_losses.append(avg_loss)

    print(
        f"CNN Epoch {epoch+1:02d}/{CNN_EPOCHS} "
        f"| Loss: {avg_loss:.6f}"
    )


# =====================================================================
# 12. A* GLOBAL ROUTING
# =====================================================================

def astar_route(
    start,
    goal,
    congestion=None,
    congestion_weight=0.0
):

    if start == goal:
        return [start]

    directions = [
        (0,1),
        (0,-1),
        (1,0),
        (-1,0)
    ]

    def heuristic(a):

        return (
            abs(a[0]-goal[0])
            + abs(a[1]-goal[1])
        )

    queue = [(heuristic(start), start)]

    came_from = {}

    cost_so_far = {start: 0.0}

    while queue:

        _, current = heapq.heappop(queue)

        if current == goal:

            path = [current]

            while current in came_from:

                current = came_from[current]

                path.append(current)

            return path[::-1]

        for dx, dy in directions:

            nxt = (
                current[0]+dx,
                current[1]+dy
            )

            x, y = nxt

            if not (
                0 <= x < GRID_SIZE
                and 0 <= y < GRID_SIZE
            ):
                continue

            congestion_cost = 0.0

            if congestion is not None:

                congestion_cost = (
                    congestion_weight
                    * float(congestion[y,x])
                )

            new_cost = (
                cost_so_far[current]
                + 1.0
                + congestion_cost
            )

            if (
                nxt not in cost_so_far
                or new_cost < cost_so_far[nxt]
            ):

                cost_so_far[nxt] = new_cost

                priority = (
                    new_cost + heuristic(nxt)
                )

                heapq.heappush(
                    queue,
                    (priority, nxt)
                )

                came_from[nxt] = current

    return []


# =====================================================================
# 13. ROUTE ALL NETS
# =====================================================================

def perform_global_routing(
    positions,
    edges,
    congestion=None,
    congestion_weight=0.0
):

    usage = np.zeros(
        (GRID_SIZE, GRID_SIZE),
        dtype=np.float32
    )

    total_wirelength = 0

    successful = 0

    for a, b in edges:

        start = (
            int(np.clip(
                positions[a,0]*(GRID_SIZE-1),
                0, GRID_SIZE-1
            )),
            int(np.clip(
                positions[a,1]*(GRID_SIZE-1),
                0, GRID_SIZE-1
            ))
        )

        goal = (
            int(np.clip(
                positions[b,0]*(GRID_SIZE-1),
                0, GRID_SIZE-1
            )),
            int(np.clip(
                positions[b,1]*(GRID_SIZE-1),
                0, GRID_SIZE-1
            ))
        )

        path = astar_route(
            start,
            goal,
            congestion,
            congestion_weight
        )

        if path:

            successful += 1

            total_wirelength += len(path)-1

            for x, y in path:

                usage[y,x] += 1

    overflow_map = np.maximum(
        usage - ROUTING_CAPACITY,
        0
    )

    return {
        "wirelength": float(total_wirelength),
        "usage": usage,
        "overflow": float(overflow_map.sum()),
        "max_overflow": float(overflow_map.max()),
        "successful_routes": successful
    }


# =====================================================================
# 14. PERFORMANCE METRICS
# =====================================================================

def calculate_hpwl(positions, edges):

    total = 0.0

    for a, b in edges:

        total += abs(
            float(positions[a,0]-positions[b,0])
        )

        total += abs(
            float(positions[a,1]-positions[b,1])
        )

    return total


def calculate_mae(predicted, actual):

    return float(
        np.mean(
            np.abs(predicted-actual)
        )
    )


def calculate_mse(predicted, actual):

    return float(
        np.mean(
            (predicted-actual)**2
        )
    )


def calculate_correlation(actual, predicted):

    a = actual.flatten()
    b = predicted.flatten()

    if np.std(a) < 1e-12 or np.std(b) < 1e-12:

        return 0.0

    return float(
        np.corrcoef(a,b)[0,1]
    )


def calculate_displacement(initial, final):

    distances = np.sqrt(
        np.sum(
            (initial-final)**2,
            axis=1
        )
    )

    return float(np.mean(distances))


def calculate_density_peak(positions):

    grid = np.zeros(
        (10,10),
        dtype=np.float32
    )

    for x, y in positions:

        gx = min(int(x*10),9)

        gy = min(int(y*10),9)

        grid[gy,gx] += 1

    return float(grid.max())


def percentage_reduction(baseline, proposed):

    if abs(baseline) < 1e-12:
        return 0.0

    return (
        100*(baseline-proposed)/abs(baseline)
    )


# =====================================================================
# 15. GENERATE FIVE TEST CIRCUITS
# =====================================================================

print("\nGenerating five test circuits...")

test_circuits = [
    generate_circuit()
    for _ in range(TEST_SAMPLES)
]

results = []

placement_data = []

congestion_data = []

routing_data = []

# =====================================================================
# 16. RUN COMPLETE FIVE-CIRCUIT EXPERIMENT
# =====================================================================

graph_model.eval()

cnn_model.eval()

print("\n" + "="*75)

print("STARTING FIVE-CIRCUIT EVALUATION")

print("="*75)

for idx, circuit in enumerate(test_circuits):

    print(
        f"\nProcessing Circuit {idx+1}/{TEST_SAMPLES}"
    )

    initial = circuit["initial"]

    truth = circuit["true"]

    edges = circuit["edges"]

    features = torch.tensor(
        circuit["features"][None],
        dtype=torch.float32,
        device=DEVICE
    )

    adjacency = torch.tensor(
        circuit["adjacency"][None],
        dtype=torch.float32,
        device=DEVICE
    )

    # --------------------------------------------------------------
    # Graph Transformer Prediction
    # --------------------------------------------------------------

    with torch.no_grad():

        graph_prediction = graph_model(
            features,
            adjacency
        ).cpu().numpy()[0]

    # --------------------------------------------------------------
    # Diffusion-Inspired Placement
    # --------------------------------------------------------------

    refined = diffusion_refinement(
        graph_prediction,
        initial
    )

    # --------------------------------------------------------------
    # Congestion Map
    # --------------------------------------------------------------

    initial_demand = generate_demand_map(
        initial, edges
    )

    true_demand = generate_demand_map(
        truth, edges
    )

    # --------------------------------------------------------------
    # CNN Congestion Prediction
    # --------------------------------------------------------------

    demand_tensor = torch.tensor(
        initial_demand[None,None],
        dtype=torch.float32,
        device=DEVICE
    )

    with torch.no_grad():

        prediction = cnn_model(
            demand_tensor
        ).cpu().numpy()[0,0]

    prediction = np.maximum(
        prediction, 0
    )

    # --------------------------------------------------------------
    # Congestion Evaluation
    # --------------------------------------------------------------

    mse = calculate_mse(
        true_demand, prediction
    )

    mae = calculate_mae(
        true_demand, prediction
    )

    corr = calculate_correlation(
        true_demand, prediction
    )

    # --------------------------------------------------------------
    # Baseline Routing
    # --------------------------------------------------------------

    baseline_route = perform_global_routing(
        initial,
        edges
    )

    # --------------------------------------------------------------
    # Proposed Congestion-Aware Routing
    # --------------------------------------------------------------

    proposed_route = perform_global_routing(
        refined,
        edges,
        prediction,
        congestion_weight=0.15
    )

    # --------------------------------------------------------------
    # Placement Evaluation
    # --------------------------------------------------------------

    baseline_hpwl = calculate_hpwl(
        initial, edges
    )

    proposed_hpwl = calculate_hpwl(
        refined, edges
    )

    placement_mae = calculate_mae(
        refined, truth
    )

    displacement = calculate_displacement(
        initial, refined
    )

    baseline_density = calculate_density_peak(
        initial
    )

    proposed_density = calculate_density_peak(
        refined
    )

    # --------------------------------------------------------------
    # Store Results
    # --------------------------------------------------------------

    results.append({

        "Circuit": f"Circuit {idx+1}",

        "Baseline HPWL": baseline_hpwl,

        "Proposed HPWL": proposed_hpwl,

        "HPWL Reduction (%)": percentage_reduction(
            baseline_hpwl,
            proposed_hpwl
        ),

        "Placement MAE": placement_mae,

        "Average Displacement": displacement,

        "Baseline Density Peak": baseline_density,

        "Proposed Density Peak": proposed_density,

        "Density Reduction (%)": percentage_reduction(
            baseline_density,
            proposed_density
        ),

        "Congestion MSE": mse,

        "Congestion MAE": mae,

        "Congestion Correlation": corr,

        "Baseline Routing Wirelength": baseline_route["wirelength"],

        "Proposed Routing Wirelength": proposed_route["wirelength"],

        "Baseline Routing Overflow": baseline_route["overflow"],

        "Proposed Routing Overflow": proposed_route["overflow"],

        "Successfully Routed Nets": proposed_route["successful_routes"],

        "Total Nets": len(edges),

        "Routing Success (%)": (
            100 * proposed_route["successful_routes"] / len(edges)
        )

    })

    placement_data.append(
        (initial.copy(), truth.copy(), refined.copy())
    )

    congestion_data.append(
        (
            initial_demand.copy(),
            true_demand.copy(),
            prediction.copy()
        )
    )

    routing_data.append(
        (
            baseline_route["usage"].copy(),
            proposed_route["usage"].copy()
        )
    )

    print("Baseline HPWL:", round(baseline_hpwl,4))

    print("Proposed HPWL:", round(proposed_hpwl,4))

    print("Placement MAE:", round(placement_mae,5))

    print("Congestion MSE:", round(mse,5))

    print("Congestion MAE:", round(mae,5))

    print("Congestion Correlation:", round(corr,5))

    print(
        "Baseline Routing:",
        baseline_route["wirelength"]
    )

    print(
        "Proposed Routing:",
        proposed_route["wirelength"]
    )

    print(
        "Routing Success:",
        proposed_route["successful_routes"],
        "/",
        len(edges)
    )


# =====================================================================
# 17. RESULTS TABLE
# =====================================================================

results_df = pd.DataFrame(results)

results_path = os.path.join(
    OUTPUT_DIR,
    "All_Circuit_Results.csv"
)

results_df.to_csv(
    results_path,
    index=False
)

print("\n" + "="*75)

print("FIVE-CIRCUIT PERFORMANCE RESULTS")

print("="*75)

display(
    results_df.round(4)
)


# =====================================================================
# 18. AVERAGE PERFORMANCE TABLE
# =====================================================================

numeric_columns = results_df.select_dtypes(
    include=[np.number]
).columns

average_df = pd.DataFrame({

    "Metric": numeric_columns,

    "Average": [
        results_df[c].mean()
        for c in numeric_columns
    ],

    "Standard Deviation": [
        results_df[c].std()
        for c in numeric_columns
    ]

})

average_df.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "Average_Performance.csv"
    ),
    index=False
)

print("\nAVERAGE PERFORMANCE")

display(
    average_df.round(5)
)


# =====================================================================
# 19. TRAINING LOSS VISUALIZATION
# =====================================================================

plt.figure(figsize=(9,5))

plt.plot(
    graph_losses,
    marker="o",
    label="Graph Transformer"
)

plt.title(
    "Graph Transformer Training Loss"
)

plt.xlabel("Epoch")

plt.ylabel("MSE Loss")

plt.legend()

plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "Graph_Transformer_Loss.png"
    ),
    dpi=160
)

plt.show()


plt.figure(figsize=(9,5))

plt.plot(
    cnn_losses,
    marker="o",
    label="CNN Congestion Predictor"
)

plt.title(
    "CNN Training Loss"
)

plt.xlabel("Epoch")

plt.ylabel("MSE Loss")

plt.legend()

plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "CNN_Training_Loss.png"
    ),
    dpi=160
)

plt.show()


# =====================================================================
# 20. PLACEMENT VISUALIZATION FOR ALL FIVE CIRCUITS
# =====================================================================

for i, data in enumerate(placement_data):

    initial, truth, refined = data

    plt.figure(figsize=(8,6))

    plt.scatter(
        initial[:,0],
        initial[:,1],
        marker="x",
        label="Initial Placement"
    )

    plt.scatter(
        truth[:,0],
        truth[:,1],
        marker="o",
        label="Reference Placement"
    )

    plt.scatter(
        refined[:,0],
        refined[:,1],
        marker="^",
        label="Proposed Placement"
    )

    plt.title(
        f"Placement Comparison - Circuit {i+1}"
    )

    plt.xlabel("X Coordinate")

    plt.ylabel("Y Coordinate")

    plt.legend()

    plt.grid(alpha=0.3)

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            OUTPUT_DIR,
            f"Placement_Circuit_{i+1}.png"
        ),
        dpi=160
    )

    plt.show()


# =====================================================================
# 21. CONGESTION VISUALIZATION
# =====================================================================

for i, data in enumerate(congestion_data):

    initial_map, true_map, prediction = data

    fig, axes = plt.subplots(
        1,3,figsize=(15,5)
    )

    images = [
        initial_map,
        true_map,
        prediction
    ]

    titles = [
        "Initial Routing Demand",
        "Reference Routing Demand",
        "CNN Predicted Congestion"
    ]

    for ax, image, title in zip(
        axes, images, titles
    ):

        im = ax.imshow(
            image,
            origin="lower"
        )

        ax.set_title(title)

        ax.set_xlabel("X Grid")

        ax.set_ylabel("Y Grid")

        fig.colorbar(
            im, ax=ax
        )

    plt.suptitle(
        f"Congestion Prediction - Circuit {i+1}"
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            OUTPUT_DIR,
            f"Congestion_Circuit_{i+1}.png"
        ),
        dpi=160
    )

    plt.show()


# =====================================================================
# 22. ROUTING VISUALIZATION
# =====================================================================

for i, data in enumerate(routing_data):

    baseline_map, proposed_map = data

    fig, axes = plt.subplots(
        1,2,figsize=(12,5)
    )

    im1 = axes[0].imshow(
        baseline_map,
        origin="lower"
    )

    axes[0].set_title(
        "Baseline Routing Usage"
    )

    fig.colorbar(
        im1, ax=axes[0]
    )

    im2 = axes[1].imshow(
        proposed_map,
        origin="lower"
    )

    axes[1].set_title(
        "Proposed Congestion-Aware Routing"
    )

    fig.colorbar(
        im2, ax=axes[1]
    )

    plt.suptitle(
        f"Global Routing - Circuit {i+1}"
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            OUTPUT_DIR,
            f"Routing_Circuit_{i+1}.png"
        ),
        dpi=160
    )

    plt.show()


# =====================================================================
# 23. BASELINE VS PROPOSED PERFORMANCE
# =====================================================================

comparison_metrics = [

    (
        "Baseline HPWL",
        "Proposed HPWL",
        "HPWL Comparison"
    ),

    (
        "Baseline Routing Wirelength",
        "Proposed Routing Wirelength",
        "Routing Wirelength Comparison"
    ),

    (
        "Baseline Routing Overflow",
        "Proposed Routing Overflow",
        "Routing Overflow Comparison"
    ),

    (
        "Baseline Density Peak",
        "Proposed Density Peak",
        "Placement Density Comparison"
    )

]

for base_col, proposed_col, title in comparison_metrics:

    plt.figure(figsize=(9,5))

    x = np.arange(TEST_SAMPLES)

    width = 0.35

    plt.bar(
        x-width/2,
        results_df[base_col],
        width,
        label="Baseline"
    )

    plt.bar(
        x+width/2,
        results_df[proposed_col],
        width,
        label="HeteroDiffRoute-Net"
    )

    plt.xticks(
        x,
        results_df["Circuit"]
    )

    plt.title(title)

    plt.ylabel("Metric Value")

    plt.xlabel("Test Circuits")

    plt.legend()

    plt.grid(
        axis="y",
        alpha=0.3
    )

    plt.tight_layout()

    filename = (
        title.replace(" ","_")+".png"
    )

    plt.savefig(
        os.path.join(
            OUTPUT_DIR,
            filename
        ),
        dpi=160
    )

    plt.show()


# =====================================================================
# 24. CORRELATION ANALYSIS
# =====================================================================

correlation_columns = [

    "Proposed HPWL",

    "Placement MAE",

    "Average Displacement",

    "Congestion MSE",

    "Congestion MAE",

    "Congestion Correlation",

    "Proposed Routing Wirelength",

    "Proposed Routing Overflow"

]

correlation_matrix = results_df[
    correlation_columns
].corr()

print("\nPERFORMANCE CORRELATION MATRIX")

display(
    correlation_matrix.round(3)
)

plt.figure(figsize=(11,8))

im = plt.imshow(
    correlation_matrix,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.colorbar(im)

plt.xticks(
    range(len(correlation_columns)),
    correlation_columns,
    rotation=70,
    ha="right"
)

plt.yticks(
    range(len(correlation_columns)),
    correlation_columns
)

plt.title(
    "HeteroDiffRoute-Net Performance Correlation"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "Performance_Correlation.png"
    ),
    dpi=160
)

plt.show()


# =====================================================================
# 25. FINAL PERFORMANCE SUMMARY
# =====================================================================

print("\n")

print("="*75)

print("HETERODIFFROUTE-NET FINAL PERFORMANCE SUMMARY")

print("="*75)

summary_metrics = [

    "HPWL Reduction (%)",

    "Placement MAE",

    "Average Displacement",

    "Congestion MSE",

    "Congestion MAE",

    "Congestion Correlation",

    "Routing Success (%)"

]

for metric in summary_metrics:

    print(
        f"{metric:35s}: "
        f"{results_df[metric].mean():.6f}"
    )


# =====================================================================
# 26. SAVE MODEL WEIGHTS
# =====================================================================

torch.save(
    graph_model.state_dict(),
    os.path.join(
        OUTPUT_DIR,
        "Graph_Transformer_Model.pth"
    )
)

torch.save(
    cnn_model.state_dict(),
    os.path.join(
        OUTPUT_DIR,
        "Congestion_CNN_Model.pth"
    )
)


# =====================================================================
# 27. EXPORT COMPLETE EXPERIMENTAL RESULTS
# =====================================================================

zip_path = (
    "/content/HeteroDiffRouteNet_Complete_Results.zip"
)

with zipfile.ZipFile(
    zip_path,
    "w",
    zipfile.ZIP_DEFLATED
) as zf:

    for root, dirs, files_list in os.walk(
        OUTPUT_DIR
    ):

        for filename in files_list:

            filepath = os.path.join(
                root,
                filename
            )

            zf.write(
                filepath,
                arcname=os.path.relpath(
                    filepath,
                    OUTPUT_DIR
                )
            )


print("\n" + "="*75)

print("EXPERIMENT COMPLETED SUCCESSFULLY")

print("="*75)

print("All five circuits processed.")

print("Results directory:", OUTPUT_DIR)

print("Complete ZIP:", zip_path)

print("\nGenerated files include:")

for filename in sorted(os.listdir(OUTPUT_DIR)):

    print(" -", filename)


# =====================================================================
# 28. DOWNLOAD RESULTS
# =====================================================================

try:

    from google.colab import files

    files.download(zip_path)

except Exception:

    print(
        "\nDownload the ZIP from the Colab Files panel:"
    )

    print(zip_path)


print("\nHETERODIFFROUTE-NET EXECUTION FINISHED.")

print(
    "\nResearch note: This is a synthetic proof-of-concept. "
    "Official benchmark accuracy and physical-design improvements "
    "must be established using real circuit benchmark data."
)

# =====================================================================
# END OF COMPLETE IMPLEMENTATION
# =====================================================================
```
